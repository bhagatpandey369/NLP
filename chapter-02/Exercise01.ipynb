# Exercise 1: Word Frequency Counter

## Objective
Count how many times each word appears in a sentence.

## Input

```python
text = "I like NLP I like Python NLP is fun"
```

## Expected Output

```text
I : 2
like : 2
NLP : 2
Python : 1
is : 1
fun : 1
```

## Tasks

- Split the sentence into words.
- Count the frequency of each word.
- Store the result in a dictionary.
- Print the dictionary.

## Concepts Learned

- Strings
- Lists
- Dictionary
- Term Frequency (TF)

````python
class WordFrequencyCounter:

    def __init__(self, text):
        self.text = text

    def preprocess(self):
        return self.text.split()

    def count_words(self):
        words = self.preprocess()
        frequency = {}

        for word in words:
            if word in frequency:
                frequency[word] += 1
            else:
                frequency[word] = 1
        return frequency

    def display(self):
        frequency = self.count_words()
        for word, count in frequency.items():
            print(f"{word}: {count}")
````

````python
# Test Case 1

text1 = text1 = "I like NLP I like Python NLP is fun, and i like to doing math"

counter = WordFrequencyCounter(text1)
counter.display()
````

````python
# test case 2

text2 = "Machine learning is fun Machine learning is powerful"

counter2 = WordFrequencyCounter(text2)
counter2.display()
````
