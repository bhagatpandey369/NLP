**Library used:** spaCy (`en_core_web_sm`)


## 1. Import libraries

```python
import spacy
import pandas as pd

nlp = spacy.load("en_core_web_sm")
print("spaCy version:", spacy.__version__)
```

## 2. Load the news article

```python
ARTICLE_PATH = "chapter-01/tiger.txt"

with open(ARTICLE_PATH, "r", encoding="utf-8") as f:
    text = f.read()

print(f"Loaded article with {len(text.split())} words.\n")
print(text[:400], "...")
```

## 3. Perform Named Entity Recognition

```python
doc = nlp(text)

# Preview a few entities
for ent in list(doc.ents)[:10]:
    print(f"{ent.text:<20} {ent.label_}")
```

## 4. Extract all entities and their types into a DataFrame

```python
entities = []
types = []

for ent in doc.ents:
    entities.append(ent.text)
    types.append(ent.label_)

df = pd.DataFrame()
df["Entity"] = entities
df["Entity_Type"] = types

print("Extracted", len(df), "named entities.")
df.head(15)
```

## 5. Save entities to CSV

```python
OUTPUT_CSV = "BhagatPandey_NER_02.csv"

df.to_csv(OUTPUT_CSV, index=False)
print(f"Saved {len(df)} rows to {OUTPUT_CSV}")
```

## 6. Summary of entity types found in the article

```python
entity_counts = df["Entity_Type"].value_counts()
print("Entity type counts:")
print(entity_counts)

label_meanings = {
    "PERSON": "People, including fictional",
    "ORG": "Companies, agencies, institutions",
    "GPE": "Countries, cities, states",
    "DATE": "Absolute or relative dates or periods",
    "LOC": "Non-GPE locations (mountains, water bodies, regions)",
    "NORP": "Nationalities, religious or political groups",
    "CARDINAL": "Numerals that do not fall under another type",
    "PERCENT": "Percentage values",
    "MONEY": "Monetary values",
    "ORDINAL": "First, second, etc.",
    "EVENT": "Named events (conferences, wars, etc.)",
    "FAC": "Buildings, airports, highways, bridges",
    "LAW": "Named documents made into laws",
    "TIME": "Times smaller than a day",
    "PRODUCT": "Objects, vehicles, foods, etc.",
    "QUANTITY": "Measurements, e.g. weight or distance",
    "WORK_OF_ART": "Titles of books, songs, etc.",
    "LANGUAGE": "Any named language",
}

print()
print("Entity type meanings (spaCy default labels found in this article):")
for label in entity_counts.index:
    meaning = label_meanings.get(label, "See spaCy documentation")
    print(" ", label, ":", meaning)

```
