```python
for i in range(len(words)):
    current = vocab.index(words[i])

    if i > 0:
        left = vocab.index(words[i-1])
        matrix[current][left] = 1

    if i < (len(words) - 1):
        right = vocab.index(words[i+1])
        matrix[current][right] = 1
```

## Exercise 5: Implement Laplace Smoothing

```python
word_counts = {
    "cat":4,
    "dog":3,
    "bird":1
    }
```

```python
word_counts
```

```python
value = word_counts.values()
value
```

```python
total_value = sum(value)
total_value
```

```python
vocab_size = len(word_counts)
vocab_size
```

```python
for word in word_counts:
    probability = (word_counts[word] + 1) / (total_value + vocab_size)
    print(word, ":", round(probability, 3)  )
```

## Exercise 6: Simple Keyboard Prediction

```python
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]
```

```python
input_text = "I like"
```

```python
bigrams = {}
```

```python
type(bigrams)
```

```python
for sentence in corpus:
    words = sentence.split()

    for i in range(len(words)-1):
        first = words[i]
        second = words[i + 1]

        if first not in bigrams:
            bigrams[first] = {}

        if second not in bigrams[first]:
            bigrams[first][second] = 0

        bigrams[first][second] += 1
    
```
