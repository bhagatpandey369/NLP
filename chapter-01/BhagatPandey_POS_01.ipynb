## 1. Import libraries

```python
import spacy
import pandas as pd

nlp = spacy.load("en_core_web_sm")
print("spaCy version:", spacy.__version__)
```

## 2. Load the news article

The article was downloaded/saved as a plain text file named `tiger.txt` in the same folder as this notebook.

```python
ARTICLE_PATH = 'chapter-01/tiger.txt'

with open(ARTICLE_PATH, "r", encoding="utf-8") as f:
    text = f.read()
```

## 3. Perform POS tagging

```python
doc = nlp(text)

```

```python
doc
```

```python
# Preview a few tokens with their POS tags
for token in list(doc)[:15]:
    print(f"{token.text:<15} {token.pos_:<10} {token.tag_}")
```

## 4. Extract all Nouns and Verbs

spaCy's coarse `pos_` attribute labels tokens as `NOUN`, `PROPN` (proper noun), `VERB`, etc.
Here we keep `NOUN`, `PROPN`, and `VERB` tokens, skipping punctuation and whitespace.

```python
target_pos = {"NOUN", "PROPN", "VERB"}

records = []
for token in doc:
    if token.pos_ in target_pos and token.is_alpha:
        records.append({
            "Word": token.text,
            "POS_Tag": token.pos_,
            "Detailed_Tag": token.tag_,
        })

df = pd.DataFrame(records)
print(f"Extracted {len(df)} noun/verb tokens.")
df.head(15)
```

## 5. Quick summary of counts by POS category

```python
summary = df["POS_Tag"].value_counts()
summary
```

## 6. Save extracted words to CSV

```python
OUTPUT_CSV = "BhagatPandey_POS_01.csv"

df[["Word", "POS_Tag"]].to_csv(OUTPUT_CSV, index=False)
print(f"Saved {len(df)} rows to {OUTPUT_CSV}")
```
